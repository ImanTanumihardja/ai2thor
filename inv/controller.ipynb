```python
import sys
sys.path.append('../')

from ai2thor.controller import Controller
import ai2thor
import matplotlib.pyplot as plt
import numpy as np
```

```python
controller = Controller(
    agentMode='neural_os',
    server_class=ai2thor.wsgi_server.WsgiServer,
    host="127.0.0.1",
    port=8200,
    start_unity=False,
) 
```

```python
controller.step("UnpausePhysicsAutoSim")
```

```python
event = controller.step(action="RotateRight")
metadata = event.metadata
print(event, event.metadata.keys())
```

```python
metadata['neuralOS']
```

```python
event = controller.step(action="RotateLeft")
metadata = event.metadata
print(event, event.metadata.keys())
```

```python
event = controller.step(action="MoveAhead")
metadata = event.metadata
print(event, event.metadata.keys())
```

```python
# Visualize the AVP fov
avp_frame = event.third_party_camera_frames[0]
plt.imshow(avp_frame)
plt.axis('off')
plt.title("AVP FOV")
plt.show()
```

```python
# Visualize the agent fov
agent_frame = event.frame
plt.imshow(agent_frame)
plt.axis('off')
plt.show()
```

```python
event = controller.step(
    action="PickupObject",
    objectId="Apple|-00.47|+01.15|+00.48",
    forceAction=False,
    manualInteract=False
)
metadata = event.metadata
print(event, event.metadata.keys())
```
